```python
#Count class distribution
no_show_count=sum(y==1)
show_count=sum(y==0)

scale_weight=show_count/no_show_count
print(scale_weight)
```

```python
from xgboost import XGBClassifier

model=XGBClassifier(
    eval_metric='logloss',
    scale_pos_weight=scale_weight
)

model.fit(x_train,y_train)
```

### Final Conclusion

- Logistic Regression struggled due to class imbalance, showing very low recall for no-show patients.
- XGBoost effectively handled class imbalance using scale_pos_weight.
- Recall for no-show patients improved significally, making the model more useful in real_world scenarios
- Model can accurately identify patients likely to miss appointments.

Business Impact:
- Enables hospitals to send targeted reminders to high-risk patients
- Reduces missed appointments and improves scheduling efficiency
- Helps minimize revenue loss and optimize resource utilization
